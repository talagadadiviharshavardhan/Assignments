*Q4*

2.EXPLAIN HOW GROUPBY() CALCULATE THE DEPARTMENT WISE AVERAGE
The groupby('Department')['Salary'].mean() statement uses the Split-Apply-Combine mechanism to group rows by distinct department names and compute the average salary for each group.  
 How groupby() calculates department-wise average :Split: Pandas divides the DataFrame into distinct groups based on unique values in the 'Department' column (HR and IT):   HR Group: [30000, 50000]   IT Group: [40000, 60000]   
 Apply: The aggregation function .mean() is applied to the 'Salary' values of each group independently:   HR Mean: $(30000 + 50000) / 2 = 40000.0$   IT Mean: $(40000 + 60000) / 2 = 50000.0$   
 Combine: Pandas merges the calculated average values back into a single, labeled Pandas Series where department names serve as the index

 3.Explain how NumPy broadcasting enables the addition of these arrays.

 NumPy automatically stretches dimensions with a size of 1 to match the dimension of the other array. Array a (3, 1) is stretched across columns to (3, 3), and array b (1, 3) is stretched down rows to (3, 3), enabling element-wise addition into a (3, 3) matrix.  
 * Difference from 1D Arrays: Adding two 1D arrays of shape (3,) performs direct position-by-position addition resulting in a 1D array of shape (3,), whereas adding 2D column (3, 1) and row (1, 3) arrays triggers outer expansion to form a (3, 3) 2D matrix.

1 Load the dataset and inspect it using head(), info(), and isnull().sum().



````python
import numpy as np
import pandas as pd

# Creating the dirty dataset as a DataFrame
data = {
    'Student_ID': ['S101', 'S102', 'S103', 'S104', 'S105', 'S106', 'S107', 'S108', 'S102', 'S109', 'S110', 'S111', 'S112', 'S113', 'S114'],
    'Student_Name': ['Asha', 'Ravi', 'Neha', 'Karan', 'Meera', 'Arjun', 'Priya', 'Rahul', 'Ravi', 'Sneha', 'Varun', 'Diya', 'Aman', 'Nisha', 'Kabir'],
    'Department': ['IT', 'it', 'I.T.', 'HR', 'HR', 'Finance', 'finance', 'IT', 'it', 'I.T.', 'HR', 'Finance', 'IT', 'HR', 'Finance'],
    'Assessment': ['Python'] * 15,
    'Marks': ['85', '78', '92', '', 'Absent', '0', '67', '75', '78', 'NA', '88 marks', '91', '', '64', '105']
}

df = pd.DataFrame(data)

# Inspection
print("--- First 5 Rows ---")
print(df.head())
print("\n--- Dataset Info ---")
print(df.info())
print("\n--- Missing Values Count ---")
print(df.isnull().sum())
````

```text

2.Standardize department names (IT, it, I.T. $\rightarrow$ IT; finance, Finance $\rightarrow$ Finance).

````python
# Mapping inconsistent department names to clean standard names
dept_map = {
    'it': 'IT',
    'I.T.': 'IT',
    'finance': 'Finance'
}

df['Department'] = df['Department'].replace(dept_map)
print("Standardized Departments:", df['Department'].unique())
````

```text

**EXPLANATION***
Replaced variations like it and I.T. with IT, and finance with Finance using replace() to ensure consistent categorical naming.

3.Identify and remove exact duplicate records.

````python
# Identify exact duplicates and remove them
duplicates_count = df.duplicated().sum()
df = df.drop_duplicates().reset_index(drop=True)

print(f"Number of exact duplicate records removed: {duplicates_count}")
````

```text

**EXPLANATION***

Identified duplicate records (S102 Ravi) using duplicated() and purged them using drop_duplicates().

4.Convert the Marks column into numeric format, correctly handling entries like "88 marks", "Absent", "NA", and empty cells.

````python
# Extract numbers from string values (e.g., '88 marks' -> 88)
df['Clean_Marks'] = df['Marks'].astype(str).str.extract(r'(\d+)')[0]

# Convert extracted values to numeric (float), non-numerics become NaN
df['Marks_Numeric'] = pd.to_numeric(df['Clean_Marks'], errors='coerce')

print(df[['Student_ID', 'Student_Name', 'Marks', 'Marks_Numeric']])
````

```text

**EXPLANATION***
Extracted numeric digits from strings like "88 marks" using regular expression str.extract(r'(\d+)') and converted them to numbers using pd.to_numeric().

5 Distinguish absent students from missing entries and state the risk of deleting missing values without investigation.

````python
# Distinguish between Absent students and missing/unentered marks
df['Status'] = np.where(df['Marks'] == 'Absent', 'Absent',
               np.where(df['Marks_Numeric'].isna(), 'Missing', 'Present'))

print(df[['Student_ID', 'Student_Name', 'Marks_Numeric', 'Status']])
````


```text
Risk of Deleting Missing Values Without Investigation:
Deleting missing values blindly introduces sample bias by removing legitimate records (such as students who were absent vs. those whose records failed to record due to a system glitch), ultimately skewing descriptive metrics like the overall mean.

6 Identify marks outside the valid range of 0–100 and decide on their handling.

````python
# Identify marks outside 0 to 100
invalid_mask = (df['Marks_Numeric'] < 0) | (df['Marks_Numeric'] > 100)
invalid_count = invalid_mask.sum()

print("Invalid Mark Records Found:")
print(df[invalid_mask][['Student_ID', 'Student_Name', 'Marks_Numeric']])

# Flagging out of range values as NaN for re-verification
df.loc[invalid_mask, 'Marks_Numeric'] = np.nan
````

```text

**EXPLANATION***

Detected out-of-range marks (e.g., 105 for Kabir) using boolean indexing and set them to NaN for further investigation.


7.Display the final cleaned DataFrame along with a summary report of records removed, missing marks remaining, and invalid marks identified.

````python
# Final dataset cleanup
cleaned_df = df[['Student_ID', 'Student_Name', 'Department', 'Assessment', 'Marks_Numeric', 'Status']].rename(
    columns={'Marks_Numeric': 'Marks'}
)

print("================ SUMMARY REPORT ================")
print(f"1. Duplicate Records Removed    : {duplicates_count}")
print(f"2. Remaining Missing Marks (NaN) : {cleaned_df['Marks'].isna().sum()}")
print(f"3. Invalid Marks Identified     : {invalid_count}")
print("================================================\n")

print("Final Cleaned DataFrame:")
print(cleaned_df)
````
