## Section A – Conceptual Understanding
### A1. Why are NumPy arrays generally preferred over plain Python lists for numerical computation in AI/ML workloads? Mention at least two concrete reasons.
NumPy arrays store elements in a single continuous block of memory with fixed data types, allowing your CPU to perform operations on whole chunks at once using SIMD instructions. Python lists store pointers scattered around memory with dynamic types, which creates a lot of memory lookup overhead during math tasks.

### A2. Explain vectorization. Why does replacing an explicit Python loop with a vectorized NumPy operation matter with large datasets?
Vectorization lets you run operations across an entire array without writing explicit Python `for` loops. It pushes the heavy lifting down to optimized C code, making computations run significantly faster and cleaner on large datasets.

### A3. For compatible 2D arrays A and B, explain A * B vs A @ B. Describe one neural-network situation for each.
`A * B` multiplies matching positions in two arrays, which is useful in neural networks for applying weight masks or element-wise activation drops. `A @ B` performs actual matrix multiplication, which is what we use in neural networks to pass inputs through layer weights ($XW + b$).

### A4. What is broadcasting in NumPy? State the general rule used to decide whether arrays of different shapes can be combined element-wise.
Broadcasting allows NumPy to work with arrays of different shapes without manually making copies of data. Two shapes are compatible if their dimensions match, or if one of the dimensions is equal to 1.

### A5. Explain flatten() vs ravel(). In a memory-constrained application processing large images, which would you prefer and why?
`flatten()` always makes a completely new copy of the array in memory, while `ravel()` just gives you a 1D view of the original memory whenever possible. In high-memory tasks like image processing, `ravel()` is preferred because it saves RAM.

### A6. For a 2D array, explain axis=0 and axis=1 in an aggregation function such as np.sum() or np.mean(). Illustrate with a small example.
`axis=0` operates vertically down columns (collapsing rows), while `axis=1` operates horizontally across rows (collapsing columns). For instance, given a 2x2 matrix `[[1, 2], [3, 4]]`, `np.sum(axis=0)` gives `[4, 6]`, and `np.sum(axis=1)` gives `[3, 7]`.

### A7. Why is it important to normalize or standardize numerical features before training most ML models? Explain normalization vs standardization.
When features have vastly different scales, machine learning models struggle to converge or end up heavily biased toward larger numbers. Normalization squeezes values into a fixed [0, 1] range, while standardization shifts values to have a mean of 0 and standard deviation of 1.

### A8. Why do functions such as np.nanmean() exist? What happens if np.mean() is used directly on data containing NaN?
Standard `np.mean()` will break and return `NaN` if even one value in the data is missing. Functions like `np.nanmean()` automatically filter out `NaN` values and compute the average using only valid numbers.

A batch of images adds a dataset size dimension to the standard 3D image format `(height, width, channels)`. For 64 RGB images sized 128x128, the exact tensor shape is `(64, 128, 128, 3)`.

### A10. Why is reproducibility important during model development and experimentation? Refer to fixing a random seed with np.random.default_rng(seed).

- You are given a raw 1D or 2D NumPy array containing NaN values.
- Describe, step by step, the NumPy operations you would apply to:
- (a) detect missing values,
- (b) handle them appropriately and justify your strategy, and
- (c) normalize or standardize the clean data before it is fed to a model.

Implement the full pipeline in code on a sample array of your own choosing, including at least one NaN.

#### Explanation:

#### (a) Detecting Missing Values
To start off, we need to inspect the array for any absent data points. Using `np.isnan()`, we can locate missing values across the dataset:
- `np.isnan(data)` yields a boolean mask with `True` wherever a `NaN` exists.
- `np.sum(np.isnan(data))` counts the exact number of missing entries in the array.

#### (b) Handling Missing Values
Before feeding this data into a model, those blank entries need to be filled. 

A common strategy is imputing missing values using their respective column averages. We use `np.nanmean()` for this step since it computes averages while skipping `NaN` entries.

This strategy works well when missing data is sparse because:
- It preserves the rest of the valid data points in that sample.
- It prevents us from dropping entire rows and losing useful information.

#### (c) Standardizing the Cleaned Dataset
Once the missing entries are imputed, we scale the features using z-score standardization via the formula:

`z = (x - mean) / standard_deviation`

When working with 2D matrices, we set `axis=0` so the mean and standard deviation are calculated down each column individually.

This transformation ensures each feature ends up with:
- A mean of 0
- A standard deviation of 1

Rescaling features to a shared baseline keeps larger number ranges from dominating model training and helps optimization algorithms converge smoothly.

````python
data = np.array([
    [10, 20, 30],
    [15, np.nan, 35],
    [20, 25, np.nan],
    [25, 30, 45]
])

# (a) Detect missing values
missing = np.isnan(data)

print("Missing value mask:")
print(missing)

print("Total missing values:", np.sum(missing))

# (b) Handle NaN values using column-wise mean
column_means = np.nanmean(data, axis=0)

# Replace NaN values with column means
clean_data = np.where(missing, column_means, data)

print("\nColumn means:")
print(column_means)

print("\nClean data:")
print(clean_data)

# (c) Standardize the clean data
mean = np.mean(clean_data, axis=0)
std = np.std(clean_data, axis=0)

standardized_data = (clean_data - mean) / std

print("\nStandardized data:")
print(np.round(standardized_data, 2))

print("\nMean after standardization:")
print(np.round(np.mean(standardized_data, axis=0), 2))

print("\nStandard deviation after standardization:")
print(np.round(np.std(standardized_data, axis=0), 2))
````

#### 1. Identify Every Bug

There are **two main bugs** in the given code.

##### Bug 1: Incorrect Shape of `w`

The shape of `X` is `(3, 2)`, which means `X` has **3 samples and 2 features**.

But `w` is:

`w = np.array([0.5, 1.0, 0.2])`

Its shape is `(3,)`.

This is incorrect because `w` must contain **one weight for each feature**.

Therefore, `w` should have shape `(2,)`:

`w = np.array([0.5, 1.0])`

##### Bug 2: Incorrect MSE Calculation

The given expression is:

`mse = np.mean(y_true - y_pred ** 2)`

This is not the correct Mean Squared Error formula.

The correct expression is:

`mse = np.mean((y_true - y_pred) ** 2)`

The error between the actual and predicted values must be calculated first, and then the error should be squared.

#### 2. Explain Using Shapes Why Each Bug Is a Problem

The shape of `X` is:

`X.shape = (3, 2)`

This means:

- `3` → number of samples
- `2` → number of features

Therefore, the weight vector `w` must have **2 values**, one for each feature.

The given weight vector is:

`w.shape = (3,)`

This is incorrect because it contains 3 weights while `X` has only 2 features.

The correct shape is:

`w.shape = (2,)`

Then the matrix multiplication works as:

`(3, 2) @ (2,) → (3,)`

So, `y_pred` will have shape `(3,)`, which matches the shape of `y_true`.

`y_true.shape = (3,)`

Therefore, the corrected shapes are:

| Variable | Shape |
|---|---|
| `X` | `(3, 2)` |
| `w` | `(2,)` |
| `X @ w` | `(3,)` |
| `y_pred` | `(3,)` |
| `y_true` | `(3,)` |

The second bug is the MSE calculation. It is not a shape problem, because `y_true` and `y_pred` have matching shapes. The problem is that the expression does not calculate the squared error correctly.

#### 3. Explain How Operator Precedence Affects the MSE Calculation

The incorrect expression is:

`mse = np.mean(y_true - y_pred ** 2)`

In Python, the power operator `**` has **higher precedence** than the subtraction operator `-`.

Therefore, Python evaluates the expression as:

`y_true - (y_pred ** 2)`

This means only `y_pred` is squared, which is **not** the correct MSE calculation.

The correct MSE formula is:

`MSE = mean((y_true - y_pred)²)`

Therefore, we must use parentheses:

`mse = np.mean((y_true - y_pred) ** 2)`
The parentheses ensure that:

1. `y_true - y_pred` is calculated first.
2. The resulting error is squared.
3. The mean of the squared errors is calculated.

Hence, parentheses are necessary to calculate MSE correctly.

#### E1. Per-Column Min–Max Normalization
Implement min-max normalization for a 2D feature matrix on a per-column (per-feature) basis, rather than using a single global min/max for the whole array. Your solution must be fully vectorized, with no explicit loops over rows or columns.



#### Explanation:
To scale each feature independently into the $[0, 1]$ range, we apply min-max normalization across individual columns using the formula:

$$\text{X\_norm} = \frac{X - \text{min}(X, \text{axis}=0)}{\text{max}(X, \text{axis}=0) - \text{min}(X, \text{axis}=0)}$$

- Specifying `axis=0` calculates the minimum and maximum values along each column (feature) rather than across the entire matrix.
- NumPy's broadcasting automatically aligns these column-wise 1D min/max results with the 2D matrix shape, applying the operation across all rows in a single vectorized step without writing `for` loops.

````python
import numpy as np

# Sample 2D feature matrix (3 samples, 2 features)
X = np.array([[10.0, 100.0], 
              [20.0, 200.0], 
              [30.0, 150.0]])

# Per-column min-max normalization using broadcasting
X_norm = (X - X.min(axis=0)) / (X.max(axis=0) - X.min(axis=0))

print("Normalized Matrix:\n", X_norm)
# Explanation: Computes column-wise minimum and maximum along axis=0 to scale each feature into [0, 1] without loops.
````

#### E2. Three-Level Score Classification
Given a 1D array of exam scores, use a single vectorized expression (no if/else statements and no explicit loops) to classify each score as:
Fail (< 40), Average (40–70 inclusive), or Excellent (> 70).
Hint: You may need to nest or combine np.where() calls.

#### Explanation:
To categorize continuous scores into three distinct performance tiers without using explicit Python `for` loops or `if/else` conditions, we use nested `np.where()` calls:

- The outer `np.where()` checks if a score is below 40. If true, it assigns `'Fail'`.
- If false, execution passes to the inner `np.where()`, which checks if the score is $\le 70$. If true, it assigns `'Average'`.
- Anything remaining above 70 defaults to `'Excellent'`.

This approach processes the entire array in a single vectorized operation.

````python
import numpy as np

# Sample 1D array of student exam scores
scores = np.array([25, 40, 68, 71, 90, 39])

# Single vectorized expression using nested np.where()
categories = np.where(scores < 40, 'Fail', 
             np.where(scores <= 70, 'Average', 'Excellent'))

print("Scores:", scores)
print("Classifications:", categories)
# Explanation: Classifies scores into Fail, Average, or Excellent using nested np.where calls without explicit loops.
````
